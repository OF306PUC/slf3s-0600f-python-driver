Al conectar el sensor al sistema fluídico es necesario considerar lo siguiente: 
1. No apretar en exceso los puertos fluídicos acoplados al sensor ya que puede generar vías de fuga.
2. Se recomienda utilizar tuberías/puertos de enetrada con un diámetro interno igual o superior al del canal interno del sensor. De lo contrario, turbulencias podrían afectar el rendimiento del sensor a caudales elevados. **Es importante mantener una condición laminar para el fluido en cuestión**
3. Posicionar el sensor de manera horizontal. Si es vertical, se genera un offset de medición (due to gravity)
4. Lo ideal es que el fluido de entrada mantenga una temperatura $\pm 3$ °C a la temperatura del sensor para un mejor rendimiento. 
5. Preferentemente, la duración de dispensación (periodo de entrega del fluido) debe ser al menos de 500 ms.


Consideraciones en la operación del sensor: 
1. Utilizar la calibración agua para soluciones salinas (suero fisiológico). 
2. El sensor tiene una alta repetibilidad con un error del 0.8~1.5%. 
3. El prinicipio de medición es calorimétrico 
4. Aire en la linea se refiere a las burbujas que pueden estar contenidas en el caudal. Estas hacen que el output de la medición sea un valor cercano a 0. 
5. El sensor entrega mediciones de la temperatura del sensor las cuales se ven afectadas por la temperatura ambiente y la del fluido como también las condiciones operacionales del sensor. 
6. Se define el punto de operación (eléctrico): $(3.3\ \mathrm{V}, 4.5 \ \mathrm{mA})$. Dado que el sensor en *idle mode* consume $0.05 \ \mathrm{mA}$, entonces la potencia destinada para el *heater* es $P_h \approx 15.2 \ \mathrm{mW}$.
